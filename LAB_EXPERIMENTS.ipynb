```python
import numpy as np

student_scores = np.array([
    [80, 75, 85, 70],
    [90, 85, 80, 75],
    [70, 80, 90, 85],
    [85, 90, 75, 80]
])

subjects = ["Math", "Science", "English", "History"]

avg = np.mean(student_scores, axis=0)

print("Average marks:", avg)
print("Subject with highest average:", subjects[np.argmax(avg)])
```

```python
import numpy as np

product_prices = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [200, 300, 400]
])

avg_price = np.mean(product_prices)

print("Average price:", avg_price)
```

```python
import numpy as np

# Columns: Bedrooms, Area, Sale Price
house_data = np.array([
    [3, 1200, 200000],
    [5, 2000, 400000],
    [4, 1500, 300000],
    [6, 2500, 500000],
    [5, 2200, 450000]
])

selected_houses = house_data[house_data[:, 0] > 4]

avg_price = np.mean(selected_houses[:, 2])

print("Average sale price:", avg_price)
```

```python
import numpy as np

sales_data = np.array([10000, 15000, 18000, 20000])

total_sales = np.sum(sales_data)

percentage_increase = (
    (sales_data[3] - sales_data[0]) / sales_data[0]
) * 100

print("Total annual sales:", total_sales)
print("Percentage increase:", percentage_increase, "%")
```

```python
import numpy as np

fuel_efficiency = np.array([25, 30, 35, 40, 45])

average = np.mean(fuel_efficiency)

model1 = 30
model2 = 40

improvement = ((model2 - model1) / model1) * 100

print("Average fuel efficiency:", average, "MPG")
print("Percentage improvement:", improvement, "%")
```

```python
prices = [100, 200, 300]
quantities = [2, 1, 3]

discount = 10
tax = 5

total = sum(prices[i] * quantities[i] for i in range(len(prices)))

discount_amount = total * discount / 100
amount = total - discount_amount

tax_amount = amount * tax / 100
final_cost = amount + tax_amount

print("Total cost:", total)
print("Discount:", discount_amount)
print("Tax:", tax_amount)
print("Final cost:", final_cost)
```
