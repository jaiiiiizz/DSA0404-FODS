```python
import numpy as np
from scipy import stats, linalg, optimize, integrate, interpolate, signal, spatial, fft, sparse, special, constants, ndimage, io

# 1. Stats
data = [10, 20, 20, 30, 40, 50]
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=True))

# 2. Linear Algebra
A = np.array([[3,2],[1,4]])
B = np.array([7,9])
solution = linalg.solve(A, B)
print("Solution:", solution)

# 3. Optimization
def objective_function(x): return x**2 + 5*x + 6
minimum = optimize.minimize(objective_function, x0=0)
print("Minimum:", minimum.x)

# 4. Integration
def f(x): return x**2
integral_value, error = integrate.quad(f, 0, 3)
print("Integral:", integral_value)

# 5. Interpolation
x_data = np.array([1,2,3,4,5])
y_data = np.array([2,4,8,16,32])
cubic_model = interpolate.CubicSpline(x_data, y_data)
print("Interpolated Value:", cubic_model(2.5))
```

```python
from scipy.spatial import distance

p1 = [1, 2]
p2 = [4, 6]

print("Euclidean Distance:", distance.euclidean(p1, p2))
print("Manhattan Distance:", distance.cityblock(p1, p2))
```

```python
import numpy as np
from scipy import signal

data = np.array([1, 3, 7, 1, 2, 6, 0, 1])

peaks, properties = signal.find_peaks(data)

print("Peak Positions:", peaks)
print("Peak Values:", data[peaks])
```

```python
from scipy import special

print("Factorial of 5:", special.factorial(5))
print("Gamma of 5:", special.gamma(5))
print("Combination 5C2:", special.comb(5, 2))
```

```python
import numpy as np
from scipy import fft

data = np.array([1, 2, 3, 4])

result = fft.fft(data)
print("FFT Result:", result)

inverse = fft.ifft(result)
print("Inverse FFT:", inverse)
```

```python
import numpy as np
from scipy import sparse

A = np.array([
    [0, 0, 3],
    [4, 0, 0],
    [0, 5, 0]
])

S = sparse.csr_matrix(A)

print("Sparse Matrix:")
print(S)

print("Dense Matrix:")
print(S.toarray())
```
